```python
from PIL import Image
from pathlib import Path
import csv


from google.colab import drive
drive.mount("/content/drive", force_remount=True)


```

```python

# ===========================
# Step 1: Define Folders
# ===========================
original_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/paper_infrence_desktop/fix_the_scale/outdata_stuff/original_pictures_outdata")
resized_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/paper_infrence_desktop/fix_the_scale/outdata_stuff/resized_pictures_outdata")

```

```python
# ===========================
# Step 2: Scan Both Folders and Log Sizes
# ===========================
image_data = []

for original_file in original_folder.iterdir():
    if original_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        resized_file = resized_folder / (original_file.stem.lower() + ".png")
        try:
            with Image.open(original_file) as orig_img:
                orig_w, orig_h = orig_img.size
            if resized_file.exists():
                with Image.open(resized_file) as res_img:
                    res_w, res_h = res_img.size
                image_data.append([
                    original_file.name, orig_w, orig_h, res_w, res_h
                ])
            else:
                print(f"Missing resized image for: {original_file.name}")
        except Exception as e:
            print(f"Error processing {original_file.name}: {e}")
```

```python

# ===========================
# Step 3: Export to CSV
# ===========================
output_csv = resized_folder / "image_dimensions_comparison.csv"
with open(output_csv, mode="w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["filename", "original_width", "original_height", "resized_width", "resized_height"])
    writer.writerows(image_data)

print(f"Image comparison CSV saved to: {output_csv}")
```
