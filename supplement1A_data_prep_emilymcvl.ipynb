```python
# ===========================
# Step 3: User-Defined Parameters
# ===========================
# Define the paths to your Google Drive folders
input_folders = [
    "/content/drive/My Drive/MacMillan Lab/Thesis Chapter 1 /raw_pictures_to_train_model/Test_Video",
    "/content/drive/My Drive/MacMillan Lab/Thesis Chapter 1 /raw_pictures_to_train_model/marshalls_data_for_training",
    "/content/drive/My Drive/MacMillan Lab/Thesis Chapter 1 /raw_pictures_to_train_model/sophie_data_for_training_march",
    "/content/drive/My Drive/MacMillan Lab/Thesis Chapter 1 /raw_pictures_to_train_model/crowding_data"
]

# Define the output folder where selected images will be stored
output_folder = "/content/drive/My Drive/MacMillan Lab/chapter_1/pictures/cricket_measurement_pictures_raw"
# Number of images to select from each folder
num_images = 50

```

```python
# ===========================
# Step 4: Main Function to Select Images
# ===========================
def select_images(input_folders, output_folder, num_images=50):
    """
    Selects a specified number of random images from each input folder
    in Google Drive and copies them into a single output folder.

    Args:
        input_folders (list of str): List of folder paths containing images.
        output_folder (str): Path to the output folder where selected images will be stored.
        num_images (int): Number of images to select from each folder (default: 50).
    """
    # Create the output folder if it doesn't exist
    output_folder_path = Path(output_folder)
    output_folder_path.mkdir(parents=True, exist_ok=True)

    for folder in input_folders:
        folder_path = Path(folder)
        if not folder_path.exists():
            print(f"Folder not found: {folder}")
            continue

        # Get all image files in the folder (adjust extensions as needed)
        image_files = [f for f in folder_path.iterdir() if f.suffix.lower() in ['.png', '.jpg', '.jpeg']]

        # Randomly sample the specified number of images (or fewer if there aren't enough images)
        sampled_images = random.sample(image_files, min(num_images, len(image_files)))

        # Copy sampled images to the single output folder
        for image in tqdm(sampled_images, desc=f"Processing {folder_path.name}"):
            # Avoid filename conflicts by prefixing the folder name
            output_name = f"{folder_path.name}_{image.name}"  # e.g., folder1_img1.jpg
            shutil.copy(image, output_folder_path / output_name)

        print(f"Selected {len(sampled_images)} images from {folder} and added to {output_folder_path}")

```

```python
# ===========================
# Step 5: Run the Script
# ===========================
select_images(input_folders, output_folder, num_images)
```

```python
# ===========================
# Step 6: last all images sizes
# ===========================
from PIL import Image
from pathlib import Path
from collections import Counter

# Path to the folder with your images
folder_path = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/pictures/cricket_measurement_pictures_raw"")

# Dictionary to store image dimensions
dimensions = []

# Loop through all images in the folder
for img_file in folder_path.iterdir():
    if img_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:  # Supported image formats
        try:
            with Image.open(img_file) as img:
                dimensions.append(img.size)  # (width, height)
        except Exception as e:
            print(f"Error reading {img_file.name}: {e}")

# Count occurrences of each dimension
dimension_counts = Counter(dimensions)

# Print the results
print(f"Total images checked: {len(dimensions)}")
print("Image size counts:")
for dim, count in dimension_counts.items():
    print(f"Size {dim}: {count} images")
```

```python
from PIL import Image, ImageOps
from pathlib import Path
import shutil

# ===========================
# Step 6: Define Paths
# ===========================
input_folder = Path("/content/drive/My Drive/MacMillan Lab/old_chapter_1/raw_pictures_to_train_model/crowding_data")
resized_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/measure_your_cricket/cricket_measurements_v1-emily-2025-01-13/measure-your-cricket/put-photos-here")
resized_folder.mkdir(parents=True, exist_ok=True)

# ===========================
# Step 7: Find the Smallest Image Dimensions
# ===========================
min_width = float('inf')
min_height = float('inf')

for img_file in input_folder.iterdir():
    if img_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        try:
            with Image.open(img_file) as img:
                width, height = img.size
                min_width = min(min_width, width)
                min_height = min(min_height, height)
        except Exception as e:
            print(f"Error reading {img_file.name}: {e}")

print(f"Smallest image size: {min_width}x{min_height}")

# ===========================
# Step 8: Resize Images and Save as Lowercase .png Files
# ===========================
target_size = (min_width, min_height)  # Use the smallest size as the target

for img_file in input_folder.iterdir():
    if img_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        try:
            with Image.open(img_file) as img:
                # Resize image preserving aspect ratio and crop if necessary
                img_resized = ImageOps.pad(img, target_size, method=Image.BICUBIC, color=(0, 0, 0))


                # Create a new filename: lowercase the original name (without extension) and add .png
                new_filename = img_file.stem.lower() + ".png"
                resized_path = resized_folder / new_filename

                # Save the resized image as PNG
                img_resized.save(resized_path, format="PNG")
                print(f"Resized and saved: {resized_path}")
        except Exception as e:
            print(f"Error resizing {img_file.name}: {e}")

print("Resizing complete. Original images are untouched.")
```

```python
# ===========================
# Step 4: Select 30 images for validation
# ===========================

# ===========================
# Step 1: Get Already Selected Images
# ===========================
from pathlib import Path

existing_images_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/pictures/cricket_measurement_pictures_raw")
existing_images = {img.name for img in existing_images_folder.iterdir() if img.suffix.lower() in [".png", ".jpg", ".jpeg"]}

print(f"Found {len(existing_images)} previously selected images. Avoiding duplicates.")






```

```python
# ===========================
# Step 2: Select Additional Images Without Duplicates
# ===========================
import random
import shutil
from tqdm import tqdm

fresh_raw_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/pictures/fresh_pictures_raw")
fresh_raw_folder.mkdir(parents=True, exist_ok=True)

num_more_images = 30  # Number of new images to select

def select_new_images(input_folders, output_folder, num_images=30):
    """Select new images, avoiding duplicates, and copy them to 'fresh_pictures_raw'."""
    for folder in input_folders:
        folder_path = Path(folder)
        if not folder_path.exists():
            print(f"Folder not found: {folder}")
            continue

        # Get all image files, excluding already selected ones
        image_files = [f for f in folder_path.iterdir() if f.suffix.lower() in ['.png', '.jpg', '.jpeg'] and f.name not in existing_images]

        if not image_files:
            print(f"No new images found in {folder}.")
            continue

        # Select new images
        sampled_images = random.sample(image_files, min(num_images, len(image_files)))

        # Copy sampled images to 'fresh_pictures_raw'
        for image in tqdm(sampled_images, desc=f"Processing {folder_path.name}"):
            output_name = f"{folder_path.name}_{image.name}"  # Prefix with folder name
            shutil.copy(image, fresh_raw_folder / output_name)

        print(f"Selected {len(sampled_images)} new images from {folder} and added to {fresh_raw_folder}")

# Run the selection process
select_new_images(input_folders, fresh_raw_folder, num_more_images)
```

```python
# ===========================
# Step 3: Resize New Images & Store in "fresh_pictures"
# ===========================
from PIL import Image, ImageOps

fresh_resized_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/pictures/fresh_pictures")
fresh_resized_folder.mkdir(parents=True, exist_ok=True)

# Find the smallest image size among the new images
min_width = float('inf')
min_height = float('inf')

for img_file in fresh_raw_folder.iterdir():
    if img_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        try:
            with Image.open(img_file) as img:
                width, height = img.size
                min_width = min(min_width, width)
                min_height = min(min_height, height)
        except Exception as e:
            print(f"Error reading {img_file.name}: {e}")

print(f"Smallest image size for new selection: {min_width}x{min_height}")

# Resize and save images
target_size = (min_width, min_height)

for img_file in fresh_raw_folder.iterdir():
    if img_file.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        try:
            with Image.open(img_file) as img:
                img_resized = ImageOps.fit(img, target_size, method=Image.BICUBIC)

                new_filename = img_file.stem.lower() + ".png"
                resized_path = fresh_resized_folder / new_filename

                img_resized.save(resized_path, format="PNG")
                print(f"Resized and saved: {resized_path}")
        except Exception as e:
            print(f"Error resizing {img_file.name}: {e}")

print("Resizing complete.")
```

```python
# ===========================
# Step 4: Copy Resized Images to Final Folder
# ===========================
final_folder = Path("/content/drive/My Drive/MacMillan Lab/chapter_1/measure_your_cricket/cricket_measurements_v1-emily-2025-01-13/labeled-data/cricket_pictures")

copied_count = 0

for resized_file in fresh_resized_folder.iterdir():
    if resized_file.suffix.lower() == ".png":
        shutil.copy(resized_file, final_folder)
        copied_count += 1
        print(f"Copied: {resized_file.name} to {final_folder}")

print(f"All resized images have been copied to: {final_folder}")
print(f"Total images copied: {copied_count}")
```
